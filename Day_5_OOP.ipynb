```python
class Car:
  def __init__(self,make,model): #method
    self.make = make
    self.model = model
    print("Good Choice!! Your car model is selected")


def show_car_info(car_obj):  #function
  return f"I am going to buy {car_obj.model} of {car_obj.make} this year."

def premium_car_collection(car_details):
  return f"I purchased {car_details.make} {car_details.model}."

my_car = Car("Mahindra","X7OO")
print(show_car_info(my_car))

my_car2 = Car("BMW","M5")
print(premium_car_collection(my_car2))

```

```python
class Car:
  def __init__(self,make,model): #special method(constructor)
    self.make = make
    self.model = model
    print("Good Choice!! Your car model is selected")


  def show_car_info(car_obj):  #normal method
    return f"I am going to buy {car_obj.model} of {car_obj.make} this year."

def premium_car_collection(car_details):
  return f"I purchased {car_details.make} {car_details.model}."

my_car = Car("Mahindra","X7OO")
show_car_info(my_car)

my_car2 = Car("BMW","M5")
premium_car_collection(my_car2)

my_car = Car("Mahindra","X7OO")
my_car.show_car_info()
```
