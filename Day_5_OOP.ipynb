```python
#Encapsulation
"""
🔒 1. Need for Encapsulation
Encapsulation is about bundling data (attributes) and behavior (methods) inside a class.

It protects internal details from outside interference and misuse.

Example: You don’t want anyone directly changing a bank account balance; instead, they should use methods like deposit() or withdraw().

🛡️ 2. Private Attributes
In Python, attributes can be made private by prefixing them with double underscores (__).

Private attributes are hidden from direct access outside the class.

class BankAccount:
    def __init__(self, owner, balance):
        self.__owner = owner      # private
        self.__balance = balance  # private

🔑 3. Getter and Setter Methods
Since private attributes can’t be accessed directly, we use getter and setter methods.

Getter → retrieves the value safely.
Setter → updates the value with validation.
"""
class BankAccount:
    def __init__(self, owner, balance):
        self.__owner = owner
        self.__balance = balance

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative!")

"""
📊 4. Class Diagram (Text-Based)
Here’s a simple diagram showing encapsulation:

Code
+-----------------------------------+
|         Class: ATM       |
+-----------------------------------+
|           Attributes              |
|  - __owner    (private)           |
|  - __balance  (private)           |
+-----------------------------------+
|            Methods                |
|  + get_balance()   (public)       |
|  + set_balance()   (public)       |
+-----------------------------------+
👉 This structure shows data hiding (private attributes) and controlled access (public methods).

"""

#Inheritance
"""
(Private attributes can not be inherit.)

🧩 1. Need for Inheritance
Inheritance allows a class (child) to reuse code from another class (parent).

It reduces duplication and improves maintainability.

Example: All vehicles have start(), but cars and bikes add their own features.
"""
class Vehicle:
    def start(self):
        print("Vehicle starts")

class Car(Vehicle):
    def drive(self):
        print("Car drives on road")

class Bike(Vehicle):
    def ride(self):
        print("Bike rides on track")
"""
🔄 2. Overriding
A child class can override a parent’s method to provide its own implementation.
"""

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):   # overriding
        print("Dog barks")

d = Dog()
d.sound()   # Output: Dog barks
"""
👉 The child’s version replaces the parent’s version.

🪜 3. Super
The super() function lets you call the parent class’s method inside the child class.

Useful when you want to extend functionality instead of replacing it.
"""
class Parent:
    def show(self):
        print("Parent method")

class Child(Parent):
    def show(self):
        super().show()   # call parent method
        print("Child method")

c = Child()
c.show()
"""
👉 Output:
Parent method
Child method

🏛️ 4. Types of Inheritance
Python supports several types:

1️⃣ Single Inheritance
One parent → one child.
"""
class A:
    def feature1(self): print("Feature 1")

class B(A):
    def feature2(self): print("Feature 2")
"""
2️⃣ Multiple Inheritance
One child → multiple parents.
"""
class Father:
    def skill1(self): print("Gardening")

class Mother:
    def skill2(self): print("Cooking")

class Child(Father, Mother):
    def skill3(self): print("Painting")
"""
3️⃣ Multilevel Inheritance
Chain of inheritance.
"""
class Grandparent:
    def feature1(self): print("House")

class Parent(Grandparent):
    def feature2(self): print("Car")

class Child(Parent):
    def feature3(self): print("Bike")
"""
4️⃣ Hierarchical Inheritance
One parent → multiple children.
"""
class Vehicle:
    def start(self): print("Starts")

class Car(Vehicle):
    def drive(self): print("Car drives")

class Bike(Vehicle):
    def ride(self): print("Bike rides")
"""
📊 Class Diagram (Text-Based)
Code
          Vehicle
             |
   ---------------------
   |                   |
   Car                Bike



Code
Grandparent → Parent → Child


Code
Father      Mother
   \         /
     \     /
      Child

🎯 Summary

Need → Code reuse, maintainability.
Overriding → Child replaces parent’s method.
Super → Call parent’s method inside child.
Types → Single, Multiple, Multilevel, Hierarchical.

"""
```

```python
#Concepts under Inheritance

#Inheriting Constructor

class Phone:
  def __init__(self,price,brand,camera):
    print("Inside phone constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

class SmartPhone(Phone):
  pass

s = SmartPhone(20000,'Apple',15)
print(s.brand)
```

```python
#Accessing Private Attributes (child class can not access parent class hidden attributes.)

class Phone:
  def __init__(self,price,brand,camera):
    print("Inside phone constructor")
    self.price = price
    self.__brand = brand   #Use double underscore to make it private.
    self.camera = camera

class SmartPhone(Phone):
  pass

s = SmartPhone(20000,'Apple',15)
print(s.__brand)
```

```python
#Polymorphism in Inheritance

class Phone:
  def __init__(self,price,brand,camera):
    print("Inside phone constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

  def buy(self):                  #buy is here
    print("Buying a phone")

class SmartPhone(Phone):
  def buy(self):                  #buy is here too
    print("Buying a smartphone")  #and output will choose to show this as output as we have called from smartphone class.

s = SmartPhone(20000,'Apple',15)
s.buy()

#Method Overriding --> (this topic comes under polymorphism, other topics under polymorphism are overloading and operator overloading.)
```
