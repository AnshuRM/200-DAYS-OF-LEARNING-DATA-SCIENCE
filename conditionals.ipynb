```python
#Check whether a number is even or odd.

num = int(input("enter a number:"))

if num%2==0:
  print("Even")
else:
  print("Odd")
```

```python
#Check whether a number is positive, negative, or zero.

num = int(input("enter a number: "))

if num>0:
  print("Positive")

elif num<0:
  print("Negative")

else:
  print("Zero")
```

```python
#Find the largest of two numbers

num1 = int(input("enter your number: "))
num2 = int(input("enter your number: "))

if num1 > num2:
  print("greater number is",num1)

else:
  print("greater number  is",num2)
```

```python
#Find the largest of three numbers.

num1 = int(input("enter your number: "))
num2 = int(input("enter your number: "))
num3 = int(input("enter your number: "))

if num1>num2:
  print("largest number is", num1 )

elif num2>num3:
  print("largest number is", num2 )

else:
  print("largest number is", num3)
```

```python
#Check whether a year is a leap year.
#if a number is divisible by 400 and if the number is divisible is by 4 then it should not be divisible by 100.

num = int(input("enter a year:"))

if num%400 == 0 or (num%100!=0 and num%4==0):
  print("It is a leap year.")

else:
  print("Not a leap year.")
```
