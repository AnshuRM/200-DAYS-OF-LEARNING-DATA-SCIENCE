```python
#Loop
"""

Loops are control structures that let you repeat a block of code multiple times. The two main types are the for loop (used to iterate over sequences like lists, strings, or ranges)
and the while loop (used when repetition depends on a condition). Control statements like break, continue, and else refine loop behavior. Nested Loops that is loops inside loops.
The inner loop runs completely for each iteration of the outer loop.

"""
```

```python
# For Loop - List iteration
# (Cleaning and Transforming Data - Data analysts often need to clean raw datasets before analysis.)

# Raw dataset: customer ages (some invalid values)
ages = [25, 42, -1, 33, 0, 19, 55]

# Clean dataset: remove invalid ages
cleaned_ages = []
for age in ages:   # iterate over list
    if age > 0:
        cleaned_ages.append(age)


print("Cleaned ages:", cleaned_ages)
```

```python
# For Loop - String iteration
# Feature Engineering for Text Analytics

# Example: customer review
review = "Great product, very useful!"

# Count vowels in the review
vowels = "aeiouAEIOU"
count = 0
for char in review:   # iterate over string
    if char in vowels:
        count += 1

print("Number of vowels in review:", count)
```

```python
# For Loop - Range

n = 5
factorial = 1
for i in range(1, n+1):
    factorial *= i
print("Factorial of", n, "is", factorial)
```

```python
# For Loop - Tuples
# Example 1: Data Visualization (Coordinates in Tuples), Analysts often store (x, y) points in tuples for plotting.

# Coordinates of data points
points = [(2, 3), (4, 5), (6, 7)]

# Iterate over tuples
for x, y in points:
    print(f"Point at X={x}, Y={y}")

#Use case: Iterating over tuples of coordinates for scatter plots or geospatial analysis.
#Real-world: Plotting customer locations, sensor readings, or time-series points.

#Example 2: Feature Engineering (Feature-Value Pairs), Tuples are handy for storing (feature, value) pairs in ML preprocessing.

# Features with values
features = [("Age", 29), ("Income", 45000), ("Experience", 5)]

# Iterate over tuples
for feature, value in features:
    print(f"{feature}: {value}")

#Use case: Iterating over tuples of features to transform or encode data.
#Real-world: Preparing structured datasets for machine learning models (e.g., converting categorical features, scaling numeric ones).
```

```python
#Loops (While Loop)
"""
+-----------------------------------+
|        Flowchart: For Loop        |
+-----------------------------------+
| Initialization Expression         |
| - Set loop variable (start value) |
| - Example: i = 0                  |
+-----------------------------------+
| Test Expression                   |
| - Check condition (True/False)    |
| - Example: i < end                |
+-----------------------------------+
|             /   \                 |
|        True      False            |
|         |          |              |
|         v          v              |
|     For Loop Body  Exit Loop      |
| - Execute statements inside loop  |
| - Example: print(i)               |
+-----------------------------------+
| Update Expression                 |
| - Increment/Decrement variable    |
| - Example: i += step              |
+-----------------------------------+
| Loop back to Test Expression      |
| - Repeat until condition is False |
+-----------------------------------+
| End of Loop                       |
+-----------------------------------+

START
Initialization expression → set loop variable

Test expression → Is condition True?
Yes → Enter loop body
For loop body → Executing with i = 1
Update expression → Increment loop variable

Test expression → Is condition True?
Yes → Enter loop body
For loop body → Executing with i = 2
Update expression → Increment loop variable

Test expression → Is condition True?
Yes → Enter loop body
For loop body → Executing with i = 3
Update expression → Increment loop variable

Test expression → Is condition True?
Yes → Enter loop body
For loop body → Executing with i = 4
Update expression → Increment loop variable

Test expression → Condition False → Exit loop
END

"""
```

```python
#Waiting for User Input

password = ""
while password != "secret":
  password = input("Enter password: ")
print("Access granted!!")
```

```python
#Counting Until a Limit

count = 0
while count < 5:
    print("Count =", count)
    count += 1
```

```python
#infinite Loop with Break

while True:
    response = input("Type 'exit' to quit: ")
    if response == "exit":
        break
    print("You typed:", response)
```

```python
"""

| Statement    | Effect                                   |
| ---          | ---                                      |
| ``break``    | Exits loop immediately                   |
| ``continue`` | Skips current iteration, continues next  |
| ``else``     | Runs only if loop ends without ``break`` |

"""
```

```python
#1.break → Exit the loop immediately
#Ends the loop regardless of the condition or sequence.
#Useful when you’ve found what you’re looking for and don’t need to continue the search.
#Example: Stop searching once a match is found


numbers = [10, 20, 30, 40, 50]
for num in numbers:
    if num == 30:
        print("Found 30, breaking out!")
        break
    print("Checking:", num)
```

```python
#2.continue → Skip current iteration, move to next
#Skips the rest of the loop body for the current iteration.
#Useful for ignoring unwanted values without stopping the loop.
#Example: Skip negative values

data = [12, -5, 8, -3, 15]
for val in data:
    if val < 0:
        continue   # skip negatives
    print("Valid value:", val)
```

```python
#3.else with loops → Runs only if loop finishes normally
#Executes after the loop ends without a break.
#Useful for search operations: confirm if item was found or not.

#Example: Search with confirmation

numbers = [1, 2, 3, 4, 5]
target = 7

for num in numbers:
    if num == target:
        print("Target found!")
        break
else:
    print("Target not found in list")
```

```python
#Example Code:-

i =0
for i in range(1, 15):       # outer loop
    for j in range(i):      # inner loop
        print(i, end=" ")
    print()

#So the role of i is: It’s the current number from the outer loop’s range.
#It controls two things:The value being printed.
#How many times the inner loop runs.

```

```python
#Nested Loop Example
#1. Matrix Operations
#Iterating over rows and columns of a dataset.

matrix = [[1, 2], [3, 4], [5, 6]]

for row in matrix:          # outer loop → rows
    for val in row:         # inner loop → columns
        print(val, end=" ")
    print()
```

```python
#Nested Loop Example
#2. Cartesian Product (Pairing Values)
#Generating all possible pairs of items.

colors = ["Red", "Blue"]
sizes = ["S", "M", "L"]

for color in colors:        # outer loop
    for size in sizes:      # inner loop
        print(color, size)
```

```python
#Nested Loop Example
#3. Nested Ranges for Patterns
#Building structured outputs (like tables).

for i in range(1, 4):       # rows
    for j in range(1, 4):   # columns
        print(f"({i},{j})", end=" ")
    print()
```
