```python
class Bank:
    def __init__(self, name):
        print("Inside __init__, id(self):", id(self))
        self.name = name

sbi = Bank("SBI")
print("Outside, id(sbi):", id(sbi))

hdfc = Bank("HDFC")
print("Outside, id(hdfc):", id(hdfc))
```

```python
"""

Self is also an object.
When we print id(self) inside the constructor, Python shows us the memory address
of the object currently being created.

For sbi:
Inside __init__, id(self) matches id(sbi) outside.
This proves that self and sbi are the same object — just two different names
pointing to the same memory location.

For hdfc:
When we create another object, self now points to hdfc.
Again, id(self) matches id(hdfc).

Key Insight:
self is not a fixed object.
It is a placeholder variable that Python automatically binds to whichever object
is being worked on.
That’s why the id(self) keeps changing depending on the object (sbi, hdfc, etc.).

✨ Takeaway
self is always the current instance.
id(self) = id(object_name) for whichever object is being operated on.
Each new object has a new memory address, so self changes accordingly.

"""
```

```python
"""
Creating fraction class using magic methods like __str__, __add__, __sub__, __mul__,
__truediv__.

website:- https://www.tutorialsteacher.com/python/magic-methods-in-python

"""

class Fraction:
  def __init__(self, n, d):
    self.num = n
    self.den = d

  def __str__(self):
    return "{}/{}".format(self.num,self.den)

  def __add__(self,other):  #using other because it will need two objects that is fractions to work on.
    final_num = other.den*self.num + self.den*other.num
    final_den = self.den*other.den
    return "{}/{}".format(final_num,final_den)

  def __sub__(self,other):
    final_num = other.den*self.num - self.den*other.num
    final_den = self.den*other.den
    return "{}/{}".format(final_num,final_den)

  def __mul__(self,other):
    final_num = self.num*other.num
    final_den = self.den*other.den
    return "{}/{}".format(final_num,final_den)

  def __truediv__(self,other):
    final_num = self.num*other.den
    final_den = self.den*other.num
    return "{}/{}".format(final_num,final_den)

a = Fraction(4,5)
b = Fraction(1,2)
print(a+b)
print()
print(a-b)
print()
print(a*b)
print()
print(a/b)
```
